# Feature Selection


#### Feature selection is a crucial step in machine learning to improve model performance, reduce overfitting, and enhance interpretability. 



#### Feature selection involves selecting the most relevant features (variables) from the dataset while discarding irrelevant or redundant ones. 
    

### SelectKBest


SelectKBest is a feature selection method from sklearn.feature_selection that selects the top k features based on a statistical score.
    
It’s widely used to reduce dimensionality and improve model performance by removing irrelevant or redundant features.

Formula: SelectKBest(score_func, k)

    score_func: Function to evaluate each feature (e.g., f_classif, chi2, mutual_info_classif)
    
    k: Number of top features to select


### Filter methods


#### f_classif (ANOVA F-value)


    Type: Supervised, for classification
    
    Method: Measures linear dependency between each feature and the target using ANOVA F-test.
    
    Requirement: Features should be continuous (numeric), target should be categorical.
    
    Pros: Fast, interpretable, standard for many datasets.
    
    Cons: Captures only linear relationships; fails on non-linear dependencies.


#### chi2 (Chi-squared test)



    Type: Supervised, for classification
    
    Method: Measures dependency between categorical features and target using Chi-squared statistics.
    
    Requirement: Features must be non-negative (0 or positive values).
    
    Pros: Good for categorical/count data like bag-of-words in NLP.
    
    Cons: Only detects linear association, cannot handle negative numbers or continuous features well.


#### mutual_info_classif (Mutual Information)



    Type: Supervised, for classification
    
    Method: Measures non-linear dependency between each feature and the target.
    
    Requirement: Works for continuous or discrete features.
    
    Pros: Can detect complex non-linear relationships that F-test cannot.
    
    Cons: Slower to compute than f_classif or chi2, less interpretable.
   

| Function                | Feature Type           | Target Type | Captures Linear/Non-linear | Pros                           | Cons                           |
| ----------------------- | ---------------------- | ----------- | -------------------------- | ------------------------------ | ------------------------------ |
| **f_classif**           | Continuous             | Categorical | Linear                     | Fast, simple                   | Misses non-linear dependencies |
| **chi2**                | Non-negative           | Categorical | Linear                     | Works for counts/categorical   | Only positive features, linear |
| **mutual_info_classif** | Continuous or discrete | Categorical | Non-linear                 | Captures complex relationships | Slower, harder to interpret    |



Rule of thumb:
    
    Use f_classif for numeric features, simple and linear.
    
    Use chi2 for count-based or categorical features (like text, one-hot).
    
    Use mutual_info_classif when you suspect non-linear relationships.

### f_classif

```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler

url = "data/iris.csv"

df = pd.read_csv(url, names=['sepal length','sepal width','petal length','petal width','target'])

features = ['sepal length', 'sepal width', 'petal length', 'petal width']

x = df.loc[:, features].values

y = df.loc[:,['target']].values

X = StandardScaler().fit_transform(x)
```

```python
X
```

```python
from sklearn.feature_selection import SelectKBest, f_classif

selector = SelectKBest(score_func=f_classif, k=2)

selector.fit(X, y.ravel())

for feature, score, selected in zip(
    features,
    selector.scores_,
    selector.get_support()
):
    print(feature, score, selected)
```

### mutual_info_classif

```python
from sklearn.feature_selection import SelectKBest, mutual_info_classif

selector = SelectKBest(
    score_func=mutual_info_classif,
    k=2
)

selector.fit(X, y.ravel())

for feature, score, selected in zip(
    features,
    selector.scores_,
    selector.get_support()
):
    print(feature, score, selected)
```

### SelectPercentile on iris data

SelectPercentile is from sklearn.feature_selection.

    It selects features based on their statistical score.
    
    Instead of k features, you choose a percentile of features, e.g., top 20% features.

    SelectPercentile(score_func=<function>, percentile=<int>)

        score_func: Function to compute feature importance (f_classif, chi2, mutual_info_classif)

        percentile: Integer (0-100) → top X% of features to select

| Method               | Selects          | Use Case                                                                 |
| -------------------- | ---------------- | ------------------------------------------------------------------------ |
| **SelectKBest**      | Top `k` features | When you know exact number of features you want                          |
| **SelectPercentile** | Top `%` features | When you want a relative number of features, independent of dataset size |


```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler

url = "data/iris.csv"

df = pd.read_csv(url, names=['sepal length','sepal width','petal length','petal width','target'])

features = ['sepal length', 'sepal width', 'petal length', 'petal width']

X = df.loc[:, features].values

y = df.loc[:,['target']].values
```

```python

from sklearn.feature_selection import SelectPercentile, chi2

selector = SelectPercentile(chi2, percentile=75)

selector.fit(X, y)

for feature, selected in zip(features, selector.get_support()):
    if selected:
        print(feature)
```

# SelectFromModel


SelectFromModel = “Keep only important features based on model weights or importance”

"SelectFromModel is an embedded feature selection technique because it uses model-derived feature importance to select features."

1. Train a model (like SVM, Logistic Regression, Random Forest)

2. The model assigns importance to each feature:
    * Coefficients (coef_) → linear models (Linear/Lasso/Ridge/Logistic Regression)
    * Feature importance (feature_importances_) → tree models (Decision Tree/Random Forest/XGBoost)

3. SelectFromModel:
    * Removes features with low importance
    * Keeps features with high importance 

```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler

url = "data/iris.csv"

df = pd.read_csv(url, names=['sepal length','sepal width','petal length','petal width','target'])

features = ['sepal length', 'sepal width', 'petal length', 'petal width']

X = df.loc[:, features].values

y = df.loc[:,['target']].values
```

```python
from sklearn.svm import SVC

from sklearn.feature_selection import SelectFromModel

model = SVC(kernel="linear")

model.fit(X, y)

selector = SelectFromModel(model, prefit=True)

for feature, selected in zip(features, selector.get_support()):
    if selected:
        print(feature)
```

```python
from sklearn.svm import SVC
from sklearn.feature_selection import SelectFromModel

model = SVC(kernel="linear")

model.fit(X, y.ravel())

selector = SelectFromModel(model, prefit=True)

for feature, selected in zip(features, selector.get_support()):
    if selected:
        print(feature)
```

### Wrapper Methods

SequentialFeatureSelector (SFS) automatically selects the best subset of features by:

    * Training a model multiple times
    
    * Adding or removing features
    
    * Choosing the combination that gives the best performance

* Forward Selection : direction='forward'

    👉 Starts with zero features → adds best ones step by step

* Backward Elimination : direction='backward'

    👉 Starts with all features → removes worst ones step by step

```python
import pandas as pd

df = pd.read_csv("paddy.csv")

y = df["Paddy yield(in Kg)"]

X = df.drop("Paddy yield(in Kg)", axis=1)

X = pd.get_dummies(X, drop_first=True)
```

```python

from sklearn.linear_model import LinearRegression

from sklearn.feature_selection import SequentialFeatureSelector

model = LinearRegression()
```

```python

selector = SequentialFeatureSelector( model, n_features_to_select=10, direction="forward", scoring="r2", cv=5)

selector.fit(X, y)
```

```python
for feature, selected in zip(X.columns, selector.get_support()):
    if selected:
        print(feature)
```

### backward

```python
from sklearn.feature_selection import SequentialFeatureSelector

selector = SequentialFeatureSelector( model, n_features_to_select=10, direction="backward", scoring="r2", cv=5)

selector.fit(X, y)
```

```python

for feature, selected in zip(X.columns, selector.get_support()):
    if selected:
        print(feature)
```

| Forward Selection        | Backward Selection             |
| ------------------------ | ------------------------------ |
| Starts with 0 features   | Starts with all features       |
| Adds best feature        | Removes worst feature          |
| Faster for many features | Better when dataset small      |
| Good for large datasets  | More computationally expensive |

#### Embedded methods 

Embedded methods perform feature selection automatically during the training of a Machine Learning model. 

Unlike filter methods (before training) and wrapper methods (after repeatedly training models), embedded methods combine model building and feature selection into a single process.


| Method                                              | ML Model                   | How Features are Selected                                 |
| --------------------------------------------------- | -------------------------- | --------------------------------------------------------- |
| **Lasso Regression (L1)**                           | Linear Regression          | Sets some coefficients to **0**, removing those features. |
| **Decision Tree**                                   | Decision Tree              | Selects features that provide the best splits.            |
| **Random Forest**                                   | Ensemble of Decision Trees | Ranks features using **feature importance** scores.       |
| **Gradient Boosting (XGBoost, LightGBM, CatBoost)** | Boosting Models            | Calculates feature importance during training.            |

### What is SelectFromModel?

SelectFromModel is a feature selection utility in scikit-learn that uses the importance or coefficients of an ML model to select the most important features.

It is called an embedded method because the feature importance comes from a trained model.

### Why is it an Embedded Method?

Because:
    
    The model is trained first.
    
    The model learns coefficients (Lasso) or feature importances (Random Forest).
    
    SelectFromModel uses those learned values to keep important features and discard the rest.

So the feature selection depends on the trained model, which makes it an embedded method.

### working with linear models and tree models for feature selection

##### linear models


A linear model is a model that makes predictions using a weighted combination of input features.
    


#### Tree Models — Feature Selection Parameters

| Model              | Feature-selection parameter |
| ------------------ | --------------------------- |
| **Linear models**  | `coef_`                     |
|                    | `abs(coef_)`                |
|                    | `C`                         |
|                    | `l1_ratio`                  |
| **Tree models**    | `feature_importances_`      |
|                    | `n_estimators`              |
|                    | `max_depth`                 |
| **Selection tool** | `threshold`                 |


```python
import pandas as pd
from sklearn.linear_model import LogisticRegression

data = pd.read_csv("data/diabetes.csv")

X = data.drop("Outcome", axis=1)
y = data["Outcome"]

model = LogisticRegression(max_iter=1000)

model.fit(X, y)

for feature, coef in zip(X.columns, model.coef_[0]):
    print(feature, coef)
```

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression

data = pd.read_csv("data/diabetes.csv")

X = data.drop("Outcome", axis=1)
y = data["Outcome"]

model = LogisticRegression(max_iter=1000)
model.fit(X, y)

# Feature importance
importance = abs(model.coef_[0])

result = pd.DataFrame({
    "Feature": X.columns,
    "Importance": importance
})

# Top 3
top3 = result.sort_values(
    "Importance",
    ascending=False
).head(5)

print(top3)
```

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

data = pd.read_csv("data/diabetes.csv")

X = data.drop("Outcome", axis=1)
y = data["Outcome"]

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X, y)

for feature, importance in zip(
    X.columns,
    model.feature_importances_
):
    print(feature, importance)
```
