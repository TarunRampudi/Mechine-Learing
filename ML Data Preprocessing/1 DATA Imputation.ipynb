```python

train.isnull().sum()
```

```python

sns.heatmap(train.isnull(),yticklabels=False,cbar=False,cmap='viridis')

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.boxplot(x='Pclass', y='Age',data=train)
plt.show()
```

```python

def impute_age(cols):
    
    Age = cols['Age']
    Pclass = cols['Pclass']

    if pd.isnull(Age):
        
        if Pclass == 1:
            return 37
        
        elif Pclass == 2:
            return 29
        
        else:
            return 24
    
    else:
        return Age
```

```python

train['Age'] = train[['Age', 'Pclass']].apply(impute_age, axis=1)
```

```python

sns.heatmap(train.isnull(),yticklabels=False,cbar=False,cmap='viridis')
plt.show()
```

### Case Study: Employee Salary Dataset

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Age': [25, 30, np.nan, 40, 35],
    'Department': ['IT', 'HR', 'IT', np.nan, 'HR'],
    'Salary': [50000, 60000, 55000, np.nan, 65000]
})

print(df)
```

```python
df.columns
```

```python
for col in df.columns:

    if df[col].dtype == "object":
        
        df[col] = df[col].fillna(df[col].mode()[0])
    
    else: 
        
        df[col] = df[col].fillna(df[col].median())
```

```python
df
```

### SKlearn

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Age': [25, 30, np.nan, 40, 35],
    'Department': ['IT', 'HR', 'IT', np.nan, 'HR'],
    'Salary': [50000, 60000, 55000, np.nan, 65000]
})

print(df)
```

### Fill Numerical Columns

For Age and Salary, use the median.

```python
from sklearn.impute import SimpleImputer

num_imputer = SimpleImputer(strategy='median')

df[['Age','Salary']] = num_imputer.fit_transform(df[['Age','Salary']])

print(df)
```

### Fill Categorical Columns

For Department, use the most frequent value.

```python
cat_imputer = SimpleImputer(strategy='most_frequent')

df[['Department']] = cat_imputer.fit_transform(df[['Department']])

print(df)
```

### KNNImputer

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Age': [25, 30, np.nan, 40, 35],
    'Department': ['IT', 'HR', 'IT', np.nan, 'HR'],
    'Salary': [50000, 60000, 55000, np.nan, 65000]
})

print(df)
```

```python
from sklearn.impute import KNNImputer

num_cols = ['Age', 'Salary']

imputer = KNNImputer()

df[num_cols] = imputer.fit_transform(df[num_cols])

print("\nAfter KNN Imputation:")
print(df)
```

### KNNImputer works only with numeric columns. 

| Strategy      | Example                                                  |
| ------------- | -------------------------------------------------------- |
| mean          | Temperature, marks, income                               |
| median        | Salary, age, house price (outliers present)              |
| most_frequent | Gender, city, department                                 |
| constant      | Unknown, Not Available, 0                                |
| KNNImputer    | When nearby records are similar and relationships matter |
